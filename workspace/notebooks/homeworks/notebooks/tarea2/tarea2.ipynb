```python
# ---------- Celda 1: importar librerías ----------
import numpy as np                     # cálculos numéricos (viene en el código de partida)
import pandas as pd                    # manejo de tablas (DataFrame)
from sklearn.datasets import load_breast_cancer  # dataset de cáncer de mama incluido en sklearn

# ---------- Celda 2: cargar y preparar los datos ----------
data = load_breast_cancer()            # trae: data.data (números), data.feature_names (nombres), data.target (etiquetas)

# Tabla: cada fila es un tumor, cada columna es una característica medida
df = pd.DataFrame(data.data, columns=data.feature_names)

# Invertimos la etiqueta:
#   original: 0 = maligno, 1 = benigno
#   con "1 - ...": 1 = maligno, 0 = benigno
df["target"] = 1 - data.target

# ---------- Celda 3: inciso a, correlación de cada característica con target ----------
# .corr() calcula la correlación de Pearson entre TODAS las columnas (matriz completa)
# ["target"] nos quedamos solo con la columna de target
correlaciones = df.corr(method="pearson")["target"]

correlaciones   # al ser la última línea de la celda, Jupyter la muestra

# ---------- Celda 4: inciso b, la mayor en valor absoluto ----------
# .drop("target") quita la fila de target (su correlación con sí misma es 1 y no nos sirve)
correlaciones = correlaciones.drop("target")

# .abs() quita el signo para comparar la FUERZA, sin importar si es positiva o negativa
# .idxmax() devuelve el NOMBRE de la característica con el valor más alto
nombre = correlaciones.abs().idxmax()

# Guardamos el valor CON signo para usarlo después
valor = correlaciones[nombre]

print("Característica:", nombre)
print("Correlación:   ", round(valor, 3))

# ---------- Celda 5: inciso c, signo e interpretación ----------
# Revisamos el signo del valor (no del valor absoluto)
if valor > 0:
    signo = "positiva"
else:
    signo = "negativa"

print("La correlación es", signo)
```
